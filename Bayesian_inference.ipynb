# Bayes theorem


### Core Concepts of Bayes' Theorem

Bayes' theorem is a fundamental concept in probability theory that describes the probability of an event based on prior knowledge of conditions that might be related to the event. Essentially, it allows you to update your beliefs in light of new evidence.

#### The Formula

$$
 P(A|B) = \frac{P(B|A) \cdot P(A)}{P(B)}
$$

#### Key Terms:
* **$P(A|B)$ (Posterior Probability):** The probability of hypothesis $A$ being true given that evidence $B$ has occurred.
* **$P(B|A)$ (Likelihood):** The probability of seeing evidence $B$ given that hypothesis $A$ is true.
* **$P(A)$ (Prior Probability):** The initial probability of hypothesis $A$ before seeing any evidence.
* **$P(B)$ (Marginal Probability):** The total probability of the evidence $B$ occurring under all possible hypotheses.

---

### Concept of Statistical Inverse

* **Main Idea:** The core power of Bayes' theorem is its ability to perform "probability inversion," allowing us to calculate the probability of a cause (hypothesis) given an effect (observed data).
* **Mathematical Flip:** It allows us to transition from $P(\text{Data} | \text{Hypothesis})$ to $P(\text{Hypothesis} | \text{Data})$, or symbolically:
  $$
  P(B|A) \implies P(A|B)
  $$
* **Forward Probability (Deduction):** Answers the question: *"What is the likelihood of observing data $B$ given that hypothesis $A$ is true?"*
* **Inverse Probability (Induction):** Answers the question: *"Given that we observed data $B$, what is the probability that hypothesis $A$ is the correct explanation?"*

## Conjugate priors & updates

$P(θ | X) ∝ P(X | θ) P(θ)$

Conjugate prior distribution will remain the same distribution when multiplied by the likelihood
If $P(θ)$ is the conjugate prior of $P(X|θ)$, the product of them would also be of the same conjugate prior distribution.
- Gaussian are conjugates with Gaussians
- Beta distribution is the conjugate prior of Normal distribution
